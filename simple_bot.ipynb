```python
import tkinter as tk

root= tk.Tk()

canvas1 = tk.Canvas(root, width = 340, height = 340)
canvas1.pack()
```

```python
def hello ():  
    label1 = tk.Label(root, text= 'first TK Project', fg='blue', font=('helvetica', 12, 'bold'))
    canvas1.create_window(150, 200, window=label1)
    
button1 = tk.Button(text='Click Me', command=hello, bg='brown',fg='white')
canvas1.create_window(150, 150, window=button1)

root.mainloop()
```
